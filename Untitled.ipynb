```python
class HeartSoundDataset(Dataset):
    def __init__(self, csv_path="training_data.csv", audio_dir="training_data", sr=2000, duration=5.0):
        self.df = pd.read_csv(csv_path)
        self.audio_dir = audio_dir
        self.sr = sr
        self.duration = duration
        self.target_length = int(sr * duration)
        
        self.id_col = 'Patient ID' if 'Patient ID' in self.df.columns else self.df.columns[0]
        self.df = self.df.dropna(subset=['Murmur'])
        
    def __len__(self):
        return len(self.df)
        
    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        patient_id = row[self.id_col]
        
        # Find matching .wav audio file for the patient
        patient_files = [f for f in os.listdir(self.audio_dir) if f.startswith(str(patient_id)) and f.endswith('.wav')]
        
        if not patient_files:
            spec_tensor = torch.zeros((1, 128, int(np.ceil(self.sr * self.duration / 512))))
            label = 0
            return spec_tensor, label
            
        audio_path = os.path.join(self.audio_dir, patient_files[0])
        
        # Load audio and pad/truncate to fixed length
        y, _ = librosa.load(audio_path, sr=self.sr)
        if len(y) < self.target_length:
            y = np.pad(y, (0, self.target_length - len(y)), 'constant')
        else:
            y = y[:self.target_length]
            
        # Convert to Mel-Spectrogram and Normalize
        mel_spec = librosa.feature.melspectrogram(y=y, sr=self.sr, n_mels=128, fmax=1000)
        log_mel_spec = librosa.power_to_db(mel_spec, ref=np.max)
        log_mel_spec = (log_mel_spec - log_mel_spec.min()) / (log_mel_spec.max() - log_mel_spec.min() + 1e-8)
        
        spec_tensor = torch.tensor(log_mel_spec, dtype=torch.float32).unsqueeze(0)
        
        # Extract label (Present = 1, Absent = 0)
        murmur_label = str(row['Murmur']).strip()
        label = 1 if murmur_label.lower() == 'present' else 0
        
        return spec_tensor, torch.tensor(label, dtype=torch.long)

# Initialize dataset and test sample count
dataset = HeartSoundDataset()
print(f"Successfully loaded {len(dataset)} samples!")
```

```python
class HeartSoundCNN(nn.Module):
    def __init__(self):
        super(HeartSoundCNN, self).__init__()
        
        self.conv_block1 = nn.Sequential(
            nn.Conv2d(1, 32, kernel_size=3, stride=1, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.MaxPool2d(2, 2)
        )
        self.conv_block2 = nn.Sequential(
            nn.Conv2d(32, 64, kernel_size=3, stride=1, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(2, 2)
        )
        self.conv_block3 = nn.Sequential(
            nn.Conv2d(64, 128, kernel_size=3, stride=1, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.MaxPool2d(2, 2)
        )
        self.global_pool = nn.AdaptiveAvgPool2d((1, 1))
        self.fc = nn.Sequential(
            nn.Flatten(),
            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(64, 2)   # Binary Classification: Normal (0) vs Murmur (1)
        )

    def forward(self, x):
        x = self.conv_block1(x)
        x = self.conv_block2(x)
        x = self.conv_block3(x)
        x = self.global_pool(x)
        x = self.fc(x)
        return x

print("Model architecture compiled successfully!")
```

```python
BATCH_SIZE = 16
EPOCHS = 10
LEARNING_RATE = 0.001

# Split dataset into 80% Train and 20% Validation
train_size = int(0.8 * len(dataset))
val_size = len(dataset) - train_size
train_dataset, val_dataset = random_split(dataset, [train_size, val_size])

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False)

print(f"Training samples: {train_size} | Validation samples: {val_size}")

# Initialize Model, Loss, and Optimizer
model = HeartSoundCNN().to(DEVICE)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE)

# Training Loop
for epoch in range(EPOCHS):
    model.train()
    running_loss = 0.0
    correct_train, total_train = 0, 0

    for spectrograms, labels in train_loader:
        spectrograms, labels = spectrograms.to(DEVICE), labels.to(DEVICE)

        optimizer.zero_grad()
        outputs = model(spectrograms)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * spectrograms.size(0)
        _, predicted = torch.max(outputs.data, 1)
        total_train += labels.size(0)
        correct_train += (predicted == labels).sum().item()

    epoch_loss = running_loss / train_size
    epoch_acc = 100 * correct_train / total_train

    # Validation Evaluation
    model.eval()
    correct_val, total_val = 0, 0
    with torch.no_grad():
        for spectrograms, labels in val_loader:
            spectrograms, labels = spectrograms.to(DEVICE), labels.to(DEVICE)
            outputs = model(spectrograms)
            _, predicted = torch.max(outputs.data, 1)
            total_val += labels.size(0)
            correct_val += (predicted == labels).sum().item()

    val_acc = 100 * correct_val / total_val if total_val > 0 else 0

    print(f"Epoch [{epoch+1}/{EPOCHS}] | Train Loss: {epoch_loss:.4f} | Train Acc: {epoch_acc:.2f}% | Val Acc: {val_acc:.2f}%")

# Save model weights
torch.save(model.state_dict(), "cardiolens_model.pth")
print("\nTraining complete! Model saved as 'cardiolens_model.pth'.")
```

```python
import os
import pandas as pd
import numpy as np
import torch
from torch.utils.data import Dataset
import librosa

class HeartSoundDataset(Dataset):
    def __init__(self, csv_path="training_data.csv", audio_dir="training_data", sr=2000, duration=5.0):
        self.df = pd.read_csv(csv_path)
        self.audio_dir = audio_dir
        self.sr = sr
        self.duration = duration
        self.target_length = int(sr * duration)
        
        self.id_col = 'Patient ID' if 'Patient ID' in self.df.columns else self.df.columns[0]
        
        # Drop rows with NaN or 'Unknown' murmurs for clean binary classification (Normal=0, Present=1)
        self.df = self.df.dropna(subset=['Murmur'])
        self.df = self.df[~self.df['Murmur'].astype(str).str.strip().str.lower().eq('unknown')]
        
        # Build a fast lookup dictionary of all .wav files recursively using os.walk
        print("Indexing audio files across directories...")
        self.file_map = {}
        for root, dirs, files in os.walk(self.audio_dir):
            for file in files:
                if file.endswith('.wav'):
                    # Extract patient ID from filename (assuming filename starts with patient id, e.g., '10004_AV.wav')
                    p_id = file.split('_')[0]
                    if p_id not in self.file_map:
                        self.file_map[p_id] = os.path.join(root, file)
        print(f"Indexed {len(self.file_map)} unique patient audio files successfully!")
        
    def __len__(self):
        return len(self.df)
        
    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        patient_id = str(row[self.id_col]).strip()
        
        # Grab audio path from our indexed map
        audio_path = self.file_map.get(patient_id, None)
        
        if not audio_path or not os.path.exists(audio_path):
            # Fallback zero tensor if file is truly missing
            spec_tensor = torch.zeros((1, 128, int(np.ceil(self.sr * self.duration / 512))))
            label = 0
            return spec_tensor, torch.tensor(label, dtype=torch.long)
            
        # Load audio and pad/truncate to fixed length
        y, _ = librosa.load(audio_path, sr=self.sr)
        if len(y) < self.target_length:
            y = np.pad(y, (0, self.target_length - len(y)), 'constant')
        else:
            y = y[:self.target_length]
            
        # Convert to Mel-Spectrogram and Normalize
        mel_spec = librosa.feature.melspectrogram(y=y, sr=self.sr, n_mels=128, fmax=1000)
        log_mel_spec = librosa.power_to_db(mel_spec, ref=np.max)
        log_mel_spec = (log_mel_spec - log_mel_spec.min()) / (log_mel_spec.max() - log_mel_spec.min() + 1e-8)
        
        spec_tensor = torch.tensor(log_mel_spec, dtype=torch.float32).unsqueeze(0)
        
        # Extract label (Present = 1, Absent = 0)
        murmur_label = str(row['Murmur']).strip()
        label = 1 if murmur_label.lower() == 'present' else 0
        
        return spec_tensor, torch.tensor(label, dtype=torch.long)

# Test the updated dataset class
dataset = HeartSoundDataset()
labels_list = [dataset[i][1].item() for i in range(len(dataset))]
from collections import Counter
print("Cleaned Dataset Label Counts (0: Normal, 1: Murmur):", Counter(labels_list))
```

```python
import torch
import librosa
import numpy as np
import torch.nn.functional as F

def predict_heart_sound(audio_path, model_path="cardiolens_model.pth", sr=2000, duration=5.0):
    """
    Loads a trained CardioLens model and predicts whether a given 
    heart sound recording contains a murmur.
    """
    # 1. Load the architecture and saved weights
    inference_model = HeartSoundCNN().to(DEVICE)
    inference_model.load_state_dict(torch.load(model_path, map_location=DEVICE))
    inference_model.eval()
    
    target_length = int(sr * duration)
    
    # 2. Load and preprocess the audio file (matching training pipeline)
    try:
        y, _ = librosa.load(audio_path, sr=sr)
    except Exception as e:
        return f"Error loading audio file: {e}"
        
    if len(y) < target_length:
        y = np.pad(y, (0, target_length - len(y)), 'constant')
    else:
        y = y[:target_length]
        
    # Convert to Log-Mel Spectrogram and normalize
    mel_spec = librosa.feature.melspectrogram(y=y, sr=sr, n_mels=128, fmax=1000)
    log_mel_spec = librosa.power_to_db(mel_spec, ref=np.max)
    log_mel_spec = (log_mel_spec - log_mel_spec.min()) / (log_mel_spec.max() - log_mel_spec.min() + 1e-8)
    
    # Convert to tensor and add batch/channel dimensions: [1, 1, 128, time_frames]
    spec_tensor = torch.tensor(log_mel_spec, dtype=torch.float32).unsqueeze(0).unsqueeze(0).to(DEVICE)
    
    # 3. Run Inference
    with torch.no_grad():
        outputs = inference_model(spec_tensor)
        probabilities = F.softmax(outputs, dim=1)
        conf, predicted = torch.max(probabilities, 1)
        
    pred_class = predicted.item()
    confidence = conf.item() * 100
    murmur_prob = probabilities[0][1].item() * 100
    
    # 4. Format Output Result
    result_label = "Murmur Present (1)" if pred_class == 1 else "Normal (0)"
    
    print(f"\n--- CardioLens Diagnostic Report ---")
    print(f"File Analyzed: {audio_path}")
    print(f"Prediction: {result_label}")
    print(f"Confidence: {confidence:.2f}%")
    print(f"Calculated Murmur Probability: {murmur_prob:.2f}%")
    
    return pred_class, murmur_prob

# --- Test the function on a sample file from your validation set ---
# Grab the first valid path from your dataset file map to test instantly:
sample_patient_id = list(dataset.file_map.keys())[0]
sample_audio_path = dataset.file_map[sample_patient_id]

print(f"Testing inference pipeline on sample patient: {sample_patient_id}")
predict_heart_sound(sample_audio_path)
```

```python
import torch
import torch.nn as nn

# Set up device (uses T4 GPU in Colab if available)
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {DEVICE}")

class HeartSoundCNN(nn.Module):
    def __init__(self):
        super(HeartSoundCNN, self).__init__()
        
        # Convolutional Feature Extractor
        self.features = nn.Sequential(
            # Block 1
            nn.Conv2d(1, 32, kernel_size=3, stride=1, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2),
            
            # Block 2
            nn.Conv2d(32, 64, kernel_size=3, stride=1, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2),
            
            # Block 3
            nn.Conv2d(64, 128, kernel_size=3, stride=1, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2)
        )
        
        # Adaptive pooling ensures consistent feature map size regardless of audio duration
        self.global_pool = nn.AdaptiveAvgPool2d((4, 4))
        
        # Classification Head
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(128 * 4 * 4, 128),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(128, 2)  # Output: 2 classes [Normal (0), Murmur Present (1)]
        )
        
    def forward(self, x):
        x = self.features(x)
        x = self.global_pool(x)
        x = self.classifier(x)
        return x

# Test the model initialization
model = HeartSoundCNN().to(DEVICE)
print("HeartSoundCNN model successfully initialized and moved to device!")
```
